## Task 5: Grouped Analysis

```python
product =df.groupby('Product Name')['Rating'].mean().sort_values(ascending=False).head()
print(product)
```

```python
df['Total Order Value']=pd.to_numeric(df['Total Order Value'],errors='coerce')
```

```python
df.groupby('Gender')['Total Order Value'].mean()
```

```python
df.groupby('Payment Method')['Total Order Value'].mean().sort_values(ascending=False).head()
```

```python
df.groupby('State')['Total Order Value'].sum().sort_values(ascending=False).head()
```

## Task 6: Correlation Heatmap

```python
sns.heatmap(df[['Discount Price','Time Spent on Website','Rating','No of Clicks']].corr(numeric_only=True) , vmin=-1,vmax=1,annot=True,cmap='coolwarm')
```
